# NOAA Global Hourly: Glasgow Airport

Download NOAA observations for Glasgow Airport, clean packed weather fields, and plot temperature and wind speed.

```python
source(file.path(getwd(), "..", "r_code", "noaa_global_hourly.R"))
```

```python
records <- fetch_global_hourly("03140099999", "2024-01-01", "2024-01-07")
message(sprintf("Downloaded %d records with %d fields.", nrow(records), ncol(records)))
head(records)
```

```python
cleaned <- clean_global_hourly(records)
head(cleaned[c("DATE", "temperature_c", "dew_point_c", "visibility_km", "wind_speed_mps")])
weather_plots <- plot_weather(cleaned, station_name = "Glasgow Airport (EGPF)")
print(weather_plots$temperature)
print(weather_plots$wind_speed)
```
